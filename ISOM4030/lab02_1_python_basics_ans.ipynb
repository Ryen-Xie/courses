#### Type Conversion

```python
num_int = 123
num_flo = 4.56
num_str = "789"

# Implicit conversion from int to float
num_new = num_int + num_flo
print("Value of num_new:", num_new)
print("datatype of num_new:", type(num_new))

# Explicit conversion from float to int
num_new = num_int + int(num_flo)
print("Value of num_new:", num_new)
print("datatype of num_new:", type(num_new))

# Explicit conversion from str to int
num_new = num_int + int(num_str)
print("Value of num_new:", num_new)
print("datatype of num_new:", type(num_new))

# Explicit conversion from str to int
num_new = str(num_int) + num_str
print("Value of num_new:", num_new)
print("datatype of num_new:", type(num_new))
```

###  Loops


#### For Loop

**Example**: Iterate over a list of sentences and count the number of words in each sentence using a for loop.



```python
sentences = ["I love Python.", "Python is versatile.", "Python makes programming fun!"]

for sentence in sentences:
    word_count = len(sentence.split())
    print(f"The sentence '{sentence}' has {word_count} words.")
```

#### Comprehension

```python
# list comprehension
squares = [num ** 2 for num in range(1, 11)]
print(squares)

# dictionary comprehension
squares_dict = {num: num ** 2 for num in range(1, 6)}
print(squares_dict)
```

#### While Loop

**Example**: Iterate over a list of names and print them in reverse order using a while loop.



```python
names = ["Alice", "Bob", "Charlie", "Dave"]
index = len(names) - 1

while index >= 0:
    print(names[index])
    index -= 1    # equivalent to: index = index - 1
```

**Example**: Create a function that takes a list of strings and returns a new list with the strings capitalized.

#### Keyword Arguments
By using keyword arguments, we can explicitly mention the argument names when calling the function. This allows us to provide the arguments in any order, making the code more readable and less prone to errors.

```python
def multiply(a, b):
    return a * b

result = multiply(a=5, b=3)
print(result)  # Output: 15
```

#### Parameter with Asterisk * & **

The calculate_sum function uses the *numbers syntax to accept an arbitrary number of positional arguments. The provided numbers are received as a tuple, allowing us to perform calculations or operations on them.



```python
def calculate_sum(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

result = calculate_sum(1, 2, 3, 4, 5)
print(result)  # Output: 15
```

The print_person_details function uses the **details syntax to accept an arbitrary number of keyword arguments. The provided arguments are received as a dictionary, allowing us to access and display their key-value pairs.



```python
def print_person_details(**details):
    for key, value in details.items():
        print(key + ":", value)

print_person_details(name="Alice", age=25, city="New York")  # must be passed by keyword arguments
# Output:
# name: Alice
# age: 25
# city: New York
```

**Exercise**: Given a dataset with input features and labels, split it into training and testing data using a predefined ratio.


```python
def split_dataset(features, labels, test_ratio):

    # Complete the program below
    total_samples = len(features)
    test_size = int(total_samples * test_ratio)
    train_size = total_samples - test_size

    # Splitting the features and labels into training and testing data
    X_train = features[:train_size]
    X_test = features[train_size:]
    y_train = labels[:train_size]
    y_test = labels[train_size:]

    return X_train, X_test, y_train, y_test

# Assuming a dataset with input features and labels
features = [[0.5, 0.2, 0.1], [0.8, 0.4, 0.9], [0.3, 0.6, 0.7], [0.6, 0.2, 0.8]]
labels = [1, 0, 1, 1]
test_ratio = 0.25

# Splitting the dataset into training and testing data
X_train, X_test, y_train, y_test = split_dataset(features, labels, test_ratio)

print("Training data - Features:", X_train)
print("Training data - Labels:", y_train)
print("Testing data - Features:", X_test)
print("Testing data - Labels:", y_test)
```

#### *lambda* Function (anonymous function)

```python
square = lambda x: x ** 2
#type(square)
result = square(4)
print(result)
```

```python
add_nums = lambda x, y: x + y
result = add_nums(5, 3)
print(result)
```

```python
data = [(3, 10), (1, 5), (2, 7)]
sorted_data = sorted(data, key=lambda x: x[1])
print(sorted_data)
```

**Examples**: Tokenize a sentence into words using a lambda function.


```python
sentence = "Natural language processing is a subfield of artificial intelligence."
tokenize_sentence = lambda s: s.split()
tokens = tokenize_sentence(sentence)
print(tokens)
```

**Example**: Sort a list of strings based on their lengths using a lambda function.

```python
words = ["machine", "learning", "natural", "language", "processing", "python", "ISOM4030"]
sorted_words = sorted(words, key=lambda w: len(w))
print(sorted_words)
```
